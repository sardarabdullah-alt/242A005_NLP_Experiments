```python
text = "i Performed a computation."
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
# Downloading Resources
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
import string
print(string.punctuation)
```

```python
text_lower=text.lower()
print(text_lower)
```

```python
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
text_clean=[]
for word in tokens:
    if word not in string.punctuation:
        text_clean.append(word)
print(text_clean)
```

```python
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok, pos='v') for tok in text_clean]
print(lemmas)
```

```python
lemmatizer.lemmatize('caught', pos='v')
```

```python
[(token, lemma) for token, lemma in zip(tokens, lemmas)]
```

```python
morphological_info = {}
for token, lemma in zip(tokens, lemmas):
  if token != lemma:
    if token.endswith("ing"):
        analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
    elif token.endswith("er" or "ation"):
        analysis = "Derivational: '{}'er person who does the action".format(lemma)
    else:
        analysis = "Inflectional or irregular form of '{}'".format(lemma)
  else:
    analysis = "Base/root form"
  morphological_info[token] = analysis
print(morphological_info)
```

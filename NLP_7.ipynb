```python
text="""U.S. recognized India as an independent country on August 15, 1947, after British rule ended."""
print(text)
```

```python
from nltk import word_tokenize, pos_tag, ne_chunk
```

```python
import nltk
nltk.download('punkt_tab')
tokens=word_tokenize(text)
print(tokens)
```

```python
tagged_tokens=pos_tag(tokens)
print(tagged_tokens)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
tagged_tokens=pos_tag(tokens)
print(tagged_tokens)
```

```python
nltk.download('tagsets_json')
nltk.help.upenn_tagset("DT")
nltk.help.upenn_tagset("VBZ")
nltk.help.upenn_tagset("VBP")
nltk.help.upenn_tagset("VBG")
```

```python
import nltk
nltk.download('maxent_ne_chunker_tab')
named_entities = ne_chunk(tagged_tokens)
print(named_entities)
```

```python
from nltk.tree import Tree

# Display named entities
for subtree in named_entities:
    if isinstance(subtree, nltk.Tree):
        entity = " ".join([word for word, tag in subtree.leaves()])
        label = subtree.label()
        print(f"Entity: {entity}, Label: {label}")
```

```python
import spacy
#Load the spacy English model
nlp = spacy.load("en_core_web_sm")

#Sample text
text = "Apple is looking at buying U.K. startup for $1 billion. Bill Gates founded Microsoft."
#Process the text
doc = nlp(text)
#Extract named entities
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```

```python
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser

# Sample sentence
sentence = "The quick brown fox jumps over the lazy dog."
sentence1 = "This is the class of natural language processing."

# Tokenization and POS tagging
tokens = word_tokenize(sentence1)
tagged = pos_tag(tokens)
```

```python
grammar_1 = "NP: {<DT>?<JJ>*<NN>}"
```

```python
from nltk.chunk import RegexpParser
chunk_parser = RegexpParser(grammar_1)
chunks = chunk_parser.parse(tagged_tokens)
print(chunks)
```

```python
from nltk.tree import tree
import svgling
svgling.draw_tree(chunks)
```

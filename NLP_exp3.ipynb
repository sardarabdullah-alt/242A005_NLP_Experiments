```python
text_lower=text.lower()
print(text_lower)
```

```python
print("Abdullah Sardar", "UIN: 242A005\n")

import re

pattern = r'https?:\/\/www\.\S+|www\.\S+|\S+\@\S+'
text_url = re.sub(pattern, "", text.lower())

print(text_url)
```

```python
import re

pattern = r'\d+\S+'
text_num = re.sub(pattern, "", text_url)

print(text_num)
```

```python
print("Abdullah Sardar", "UIN: 242A005")
!pip install emoji
```

```python
import emoji
text_emoji=emoji.replace_emoji(text_num,replace="")

print(text_emoji)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk import word_tokenize
tokenized=word_tokenize(text_emoji)
print("tokenized words",tokenized)
```

```python
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("With punctuation:", tokenized)
print()

text_punct = []

for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)

print("Without punctuation:", text_punct)
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
print((stopwords.words('english')))
```

```python
text_clean2=[]
for word in text_punct:
  if word not in stopwords.words('english'):
    text_clean2.append(word)
print(text_clean2)
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
text_stemmed=[stemmer.stem(word)for word in text_clean2]
print(text_stemmed)
```

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemma=WordNetLemmatizer()
text_lemmatized=[lemma.lemmatize(word)for word in text_clean2]
print("Lemmatized text:", text_lemmatized)
print()
print("Stemmed text:", text_stemmed)
```
